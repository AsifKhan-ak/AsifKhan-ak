```python
import random
jackpot = random.randint(1, 10)
guess = int(input("Enter your guess: "))
attemps = 1
while guess != jackpot:
    if guess >jackpot:
        print("too high")
    else:
        print("too low")
    guess = int(input("Try again"))
    attemps += 1
print(f"well done you guess {guess} right in {attemps} attempts!")
    
```
