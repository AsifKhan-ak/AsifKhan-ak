```python
total_students = int(input("Enter total students: "))
passed_students = 0
failed_students = 0

for i in range(1, total_students + 1):
    print(f"\nStudent {i}:")
    name = input("Enter student's name: ")
    roll_number = input("Enter roll number: ")
    
    prog_marks = float(input("Enter Programming marks: "))
    calc_marks = float(input("Enter Calculus marks: "))
    ict_marks = float(input("Enter ICT marks: "))
    
    total_marks = prog_marks + calc_marks + ict_marks
    percentage = (total_marks / 300) * 100
    
    if percentage >= 50:
        grade = "A" if percentage >= 80 else "B"
        status = "Pass"
        passed_students += 1
    else:
        grade = "F"
        status = "Fail"
        failed_students += 1
        
    print(f"\nresult card of:{name.upper()} ({roll_number})")
    print(f"Total Marks: {total_marks}/300")
    print(f"Percentage: {percentage:.2f}%")
    print(f"Grade: {grade}")
    print(f"Status: {status}")

# --- IS HISSE KO LOOP KE BAAHAR (UN-INDENT) KAREIN ---
print("\n" + "="*30)
print("OVERALL PERFORMANCE SUMMARY")
print(f"Total Students: {total_students}")
print(f"Passed Students: {passed_students}")
print(f"Failed Students: {failed_students}")
```
